```python
import numpy as np
import pandas as pd
```

```python
#### Train Data 

data1=pd.read_csv('../data/source_data/water_solubility_data.csv') # Dataset(A) with StdInChIKey dataset A
data2=pd.read_csv('../data/source_data/data_paper.csv',encoding='latin1') # (B)  # Dataset with Standard InChIKey and CAS_Number
data3 = pd.read_excel('../data/source_data/Supplementary_data.xlsx')#  (C) # Dataset with InChIKey
data4=pd.read_csv('../data/source_data/dataset-not-FA.csv') # (D) # Dataset with InChIKey

### Test data 
test_set=pd.read_csv('../data/source_data/dataset1-E.csv')
```

```python
data1
```

```python
# Standardize column names for consistency
data1.rename(columns={"Smiles": "SMILES", "StdInChIKey": "InChIKey"}, inplace=True)
data2.rename(columns={"Standard InChIKey": "InChIKey", "CAS Number": "CAS"}, inplace=True)
data3.rename(columns={"logS":"LogS"},inplace=True) 
data4.rename(columns={"Solubility":"LogS"},inplace=True)
test_set.rename(columns={"Solubility":"LogS"},inplace=True)
# Assign unique IDs to each dataset
data1.insert(0, 'C_ID', ['A-' + str(i + 1) for i in range(len(data1))])
data2.insert(0, 'C_ID', ['B-' + str(i + 1) for i in range(len(data2))])
data3.insert(0, 'C_ID', ['C-' + str(i + 1) for i in range(len(data3))])
data4.insert(0, 'C_ID', ['D-' + str(i + 1) for i in range(len(data4))])
test_set.insert(0, 'C_ID', ['E-' + str(i + 1) for i in range(len(test_set))])

# Include CAS Number wherever available
# If a dataset does not have CAS, add it as an empty column for consistency
if "CAS" not in data1.columns:
    data1["CAS"] = None
if "CAS" not in data2.columns:
    data2["CAS"] = None
if "CAS" not in data3.columns:
    data3["CAS"] = None    
if "CAS" not in data4.columns:
    data4["CAS"] = None
if "CAS" not in test_set.columns:
    test_set["CAS"] = None
if "Name" not in data1.columns:
    data1["Name"] = None
if "Name" not in data2.columns:
    data2["Name"] = None    
if "Name" not in data3.columns:
    data3["Name"] = None
if "Name" not in data4.columns:
    data4["Name"] = None
if "Name" not in test_set.columns:
    test_set["Name"] = None
# Select relevant columns
data1 = data1[['C_ID', 'Name','InChIKey', 'SMILES', 'CAS', 'LogS']]
data2 = data2[['C_ID','Name', 'InChIKey', 'SMILES', 'CAS', 'LogS']]
data3 = data3[['C_ID','Name', 'InChIKey', 'SMILES', 'CAS', 'LogS']]
data4 = data4[['C_ID', 'Name','InChIKey', 'SMILES', 'CAS', 'LogS']]
test_set = test_set[['C_ID','Name','InChIKey', 'SMILES', 'CAS', 'LogS', ]]
```

```python
### Size of each dataset 
print(data1.shape)
print(data2.shape)
print(data3.shape)
print(data4.shape)
print(test_set.shape)
```

```python
### Merging the train dataset to one dataframe 
frames = [data1,data2,data3,data4]
train_set = pd.concat(frames)

### Size of the train and test dataset before preprocess....
print('Size of the train dataset :',len(train_set))
print('Size of the test dataset :',len(test_set))


```

```python
train_set.reset_index(drop=True)
test_set.reset_index(drop=True)
```

```python
### In order to compare the train and test data we need to make smiles unique 
###  representation for that we have change the smiles to canonical smiles 

from rdkit.Chem import MolFromSmiles as smi2mol
from rdkit.Chem import MolToSmiles as mol2smi

## Function to create canonical smiles 
def canon(smi):
    try:
        mol=smi2mol(smi, sanitize=True)
        smi_canon=mol2smi(mol, isomericSmiles=False, canonical=True)
        return(smi_canon)
    except:
        print("ERROR")
        return(smi)
```

```python
#### Applying function to create the column with canonical smiles.  
train_set['smiles_canon'] = [canon(smi) for smi in train_set.SMILES]
test_set['smiles_canon'] = [canon(smi) for smi in test_set.SMILES]
```

```python
### Calculate  the occurence of  smiles in both train and test dataset....
train_set['occurence'] = train_set.groupby('smiles_canon')['smiles_canon'].transform('count')
test_set['occurence'] = test_set.groupby('smiles_canon')['smiles_canon'].transform('count')
```

```python
### Taking out Unique smiles from the train and test dataset 
train_set1= train_set[train_set['occurence']==1]
print(train_set1.shape)
test_set1= test_set[test_set['occurence']==1]
print(test_set1.shape)
```

```python
#### Taking out the dataframe which has duplicate smiles means more than one time occurence in the dataset ...
train_set2= train_set[train_set['occurence']>1]
print(train_set2.shape)
test_set2= test_set[test_set['occurence']>1]
print(test_set2.shape)
```

```python
grouped_df_train = train_set2.sort_values("SMILES").groupby("SMILES", sort=False).apply(lambda x: x).reset_index(drop=True)
```

```python
grouped_df_test = test_set2.sort_values("SMILES").groupby("SMILES", sort=False).apply(lambda x: x).reset_index(drop=True)
grouped_df_test
```

```python
### Not used
#Extract duplicate rows
id1 = train_set2["smiles_canon"]
train_set2=train_set2[id1.isin(id1[id1.duplicated()])].sort_values("smiles_canon")
id2 = test_set2["smiles_canon"]
test_set2=test_set2[id2.isin(id2[id2.duplicated()])].sort_values("smiles_canon")
```

```python
print(train_set2.shape)
print(test_set2.shape)
```

```python
train_set2.reset_index(drop=True)
test_set2.reset_index(drop=True)
```

```python
# Group by SMILES
grouped_train = train_set2.groupby('smiles_canon')
grouped_test = test_set2.groupby('smiles_canon')
# Filter groups where the maximum difference in solubility is less than or equal to 0.50
filtered_groups_train = grouped_train.filter(lambda x: x['LogS'].max() - x['LogS'].min() <= 0.50)
filtered_groups_test = grouped_test.filter(lambda x: x['LogS'].max() - x['LogS'].min() <= 0.50)
```

```python
print(len(filtered_groups_train))
print(len(filtered_groups_test))
```

```python
### In order to eavluate to threshold followed by the dataframe we  cross check the value of difference
grouped = filtered_groups_train.groupby('smiles_canon')['LogS'].apply(lambda x: x.sort_values().tolist())

# Calculate differences in solubility within each group
diffs = grouped.apply(lambda x: [j - i for i, j in zip(x[:-1], x[1:])])

# Create a new DataFrame with SMILES, Solubility values, and differences
result = grouped.reset_index()
result.columns = ['smiles_canon', 'Solubility_Values']

# Add the differences as a new column
result['Differences'] = diffs.values

result
```

```python
### taking the mean of duplicates smiles of the train dataset 
df_grouped_train = filtered_groups_train.groupby("smiles_canon", as_index=False).agg({
    "C_ID": "first",
    "Name":"first",
    "InChIKey": "first",
    "SMILES": "first",  # Keep the first occurrence
    "smiles_canon": "first",  # Keep the first CAS number
    "LogS": "mean",  # Take the mean of solubility values
    "CAS": "first",  # Take mean if applicable
    
})

```

```python
### taking the mean of duplicates smiles of the test dataset 
df_grouped_test = filtered_groups_test.groupby("smiles_canon", as_index=False).agg({
    "C_ID": "first",
    "Name":"first",
    "InChIKey": "first",
    "SMILES": "first",  # Keep the first occurrence
    "smiles_canon": "first",  # Keep the first CAS number
    "LogS": "mean",  # Take the mean of solubility values
    "CAS": "first",  # Take mean if applicable   
})
```

```python
### Combining the dataframe with single entry and multiple entry matching smiles and average the solubility values 

train_set=pd.concat([train_set1, df_grouped_train], axis=0, ignore_index=True)
test_set=pd.concat([test_set1, df_grouped_test], axis=0, ignore_index=True)
```

```python
print(train_set.shape)
print(test_set.shape)
```

```python
### Now dropping the occurence column 
train_set = train_set.iloc[:, :-1]
test_set = test_set.iloc[:, :-1]
```

```python
### Now we have all unique smiles , Now we need to make sure there should not be any 
### compound train data which match with test data 
Match_rows = pd.merge(test_set, train_set, on=['smiles_canon'], how='inner')
#mergedStuff.head()
print(len(Match_rows))
```

```python
### Finding the matching smiles in the train dataset..
cond = train_set['smiles_canon'].isin(Match_rows['smiles_canon'])

#### Dropping the smiles which is same and making dataframe with uniques smiles and no smiles same in ttrain  dataset 
train_set.drop(train_set[cond].index, inplace = True)
```
